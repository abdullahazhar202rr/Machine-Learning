# Python Basics

### Question 1: How do you perform basic arithmetic operations in Python?

```python
a=10
b=5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

### Question 2: How do you implement a while loop to count to five?

```python
count=1
while count<=5:
    print(count)
    count+=1
```

### Question 3: How do you use a for loop to iterate through a range of numbers?

```python
for i in range(3):
    print(i)
```

### Question 4: How do you define a simple function with no parameters?

```python
def greet():
    print("hello")
greet()
```

### Question 5: How do you define a function that accepts arguments?

```python
def add_numbers(x,y):
    return x+y
print(add_numbers(10,20))
```

### Question 6: How do you create a function with a default parameter value?

```python
def power(base,exp=2):
    return base**exp
print(power(4))
```

### Question 7: How do you use a lambda function to square a number?

```python
square=lambda x:x*x
print(square(6))
```

### Question 8: How do you append an element to a list?

```python
items=[1,2,3]
items.append(4)
print(items)
```

### Question 9: How do you remove a specific element from a list?

```python
items=[1,2,3,4]
items.remove(2)
print(items)
```

### Question 10: How do you sort a list in ascending order?

```python
nums=[5,2,8,1]
nums.sort()
print(nums)
```

### Question 11: How do you reverse the elements in a list?

```python
nums=[1,2,3]
nums.reverse()
print(nums)
```

### Question 12: How do you find the index of a specific element in a list?

```python
chars=['a','b','c']
print(chars.index('b'))
```

### Question 13: How do you clear all elements from a list?

```python
data=[1,2,3]
data.clear()
print(data)
```

### Question 14: How do you count the occurrences of an item in a list?

```python
vals=[1,2,2,3,2]
print(vals.count(2))
```

### Question 15: How do you convert two lists into a dictionary?

```python
keys=['id','name']
values=[101,'Abdullah']
result=dict(zip(keys,values))
print(result)
```

### Question 16: How do you convert a dictionary into a list of tuples?

```python
info={'a':1,'b':2}
pairs=list(info.items())
print(pairs)
```

# NumPy

### Question 1: How do you create a 1D NumPy array from a list?

```python
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr)
```

### Question 2: How do you create a 2D array (matrix) of zeros?

```python
zeros=np.zeros((3,3))
print(zeros)
```

### Question 3: How do you perform element-wise addition on two arrays?

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a+b)
```

### Question 4: How do you multiply every element in an array by a scalar?

```python
arr=np.array([10,20,30])
print(arr*2)
```

### Question 5: How do you find the square root of all elements in an array?

```python
arr=np.array([4,9,16])
print(np.sqrt(arr))
```

### Question 6: How do you calculate the mean of a NumPy array?

```python
arr=np.array([10,20,30,40])
print(np.mean(arr))
```

### Question 7: How do you create an array with a range of values?

```python
arr=np.arange(0,10,2)
print(arr)
```

### Question 8: How do you find the maximum value in a NumPy array?

```python
arr=np.array([5,12,7,18])
print(np.max(arr))
```

# Matrix Multiplication and Broadcasting

### Question 1: How do you perform matrix multiplication using the dot product?

```python
m1=np.array([[1,2],[3,4]])
m2=np.array([[5,6],[7,8]])
result=np.dot(m1,m2)
print(result)
```

### Question 2: How do you use the @ operator for matrix multiplication?

```python
m1=np.array([[1,2],[3,4]])
m2=np.array([[5,6],[7,8]])
print(m1@m2)
```

### Question 3: How do you perform matrix-vector multiplication?

```python
mat=np.array([[1,2],[3,4]])
vec=np.array([10,20])
print(np.matmul(mat,vec))
```

### Question 4: How do you add a scalar to a matrix using broadcasting?

```python
mat=np.array([[1,2],[3,4]])
print(mat+10)
```

### Question 5: How do you add a 1D array to a 2D array via broadcasting?

```python
mat=np.array([[1,2,3],[4,5,6]])
row=np.array([10,20,30])
print(mat+row)
```

### Question 6: How do you broadcast a column vector across a matrix?

```python
mat=np.array([[1,2],[3,4]])
col=np.array([[10],[20]])
print(mat+col)
```

### Question 7: How do you multiply two arrays of different shapes using broadcasting?

```python
a=np.array([1,2,3])
b=np.array([[10],[20],[30]])
print(a*b)
```

### Question 8: How do you calculate the transpose of a matrix?

```python
mat=np.array([[1,2,3],[4,5,6]])
print(mat.T)
```
